Import important libraries

```python
import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt 
```

Data Preprocessing 

```python
salary_data = pd.read_csv("D:\Python To ML All Work\ML Models\salary_data.csv")
salary_data.head()
```

```python
# Number of rows and columns in dataframes 
salary_data.shape
```

```python
# checking missing values
salary_data.isnull().sum()
```

```python
X = salary_data.iloc[:, :-1]   # Experience
Y = salary_data.iloc[:, -1]    # Salary
```

```python
print(X)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.33, random_state=2)
```

Training of linear Regression Model

```python

model = LinearRegression()

model.fit(X_train, Y_train)

test_data_pred = model.predict(X_test)
```

```python
print(test_data_pred)
```

Visualizing the predicated values and actual values

```python
plt.scatter(X_test,Y_test,color = "red" )
plt.plot(X_test,test_data_pred,color = "blue")
plt.xlabel("Work Experience")
plt.ylabel("Salary")
plt.title("Salary vs Experience")
plt.show()
```
